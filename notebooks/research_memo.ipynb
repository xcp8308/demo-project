## 初步解释

在合成数据中，数字化转型企业在 2020 年以后开始采用数字技术，DID 估计用于衡量这种转型状态与企业对数生产率之间的关系。事前平行趋势图用于检查处理组与对照组在转型前的走势是否相近，安慰剂检验用于确认处理前是否存在虚假的政策效应。Matlab 理论模型提供一个简单机制：数字投入通过改善管理与生产匹配提升生产率，但实施成本使收益呈现边际递减。
